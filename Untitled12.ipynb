```python
import pandas as pd
import numpy as np

# Creating a sample data.csv file
data = {
    'Column1': np.random.randint(0, 100, 5),
    'Column2': np.random.random(5)
}
df_sample = pd.DataFrame(data)
df_sample.to_csv('data.csv', index=False)
print('Sample data.csv created successfully.')
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load the recently created data.csv
df = pd.read_csv("data.csv")

print("Data Preview:")
print(df.head())

# Line Plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, color="blue", label="Sine Wave")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot")
plt.legend()
plt.show()

# Bar Plot
categories = ["A", "B", "C", "D"]
values = [3, 7, 5, 4]

plt.figure(figsize=(8, 4))
plt.bar(categories, values, color="orange")
plt.xlabel("Category")
plt.ylabel("Values")
plt.title("Bar Plot")
plt.show()

# Histogram
data_hist = np.random.randn(1000)

plt.figure(figsize=(8, 4))
plt.hist(data_hist, bins=20, color="purple", edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```
